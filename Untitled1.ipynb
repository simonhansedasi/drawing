```python
mu_funcs = [
    'run_simulation', 'track_species_selection', 'generate',
    'roll_stats', 'roll_3d6', 'recommend_species',
    'get_stat_combinations', 'query_species',
    'query_flexible_species', 'query_species_feats',
    'get_species_name', 'apply_species_bonus', 'get_species_id',
    'get_species_stats', 'sort_stats', 'select_class',
    'get_stat_combos', 'query_class', 'pick_background',
    'query_skills', 'query_backgrounds', 'determine_alignment'
]
```

```python
def assign_colors_to_df(df, func_col='function'):
    '''
    Adds a 'color' column to the DataFrame assigning a rainbow color to each unique function.
    '''
    unique_funcs = sorted(df[func_col].unique())
    n = len(unique_funcs)
    
    cmap = plt.get_cmap('rainbow')
    colors = [cmap(i / (n - 1 if n > 1 else 1)) for i in range(n)]
    
    # Convert RGBA to hex
    to_hex = lambda rgba: '#{:02x}{:02x}{:02x}'.format(
        int(rgba[0]*255), int(rgba[1]*255), int(rgba[2]*255)
    )
    hex_colors = list(map(to_hex, colors))
    
    # Create mapping and assign to df
    color_map = dict(zip(unique_funcs, hex_colors))
    df['color'] = df[func_col].map(color_map)
    return df, color_map
```

```python
df = df[df['function'].isin(mu_funcs)].copy()
df['time'] = np.log10(df['time'])
```

```python
# Bin the times
df['time_bin'] = pd.cut(df['time'], bins=100)

# Pivot: functions x time_bin
heatmap_df = df.pivot_table(index='function', columns='time_bin', aggfunc='size', fill_value=0)

plt.figure(figsize=(14,6))
plt.imshow(heatmap_df, aspect='auto', cmap='rainbow', origin='lower')
plt.yticks(range(len(heatmap_df.index)), heatmap_df.index)
plt.xlabel('Time bins')
plt.ylabel('Function')
plt.colorbar(label='Call count')
plt.show()
```

```python
df['cum_count'] = df.groupby('function').cumcount()

plt.figure(figsize=(12,6))
for func, g in df.groupby('function'):
    plt.plot(g['time'], g['cum_count'], color=g['color'].iloc[0], label=func)
# plt.legend()
plt.xlabel('Time')
plt.ylabel('Cumulative calls')
plt.show()
```

```python
# 1. Bin the time into discrete intervals
n_bins = 200  # adjust for resolution
time_bins = np.linspace(df['time'].min(), df['time'].max(), n_bins+1)
df['time_bin'] = np.digitize(df['time'], time_bins) - 1

# Clip any indices that are exactly n_bins
df['time_bin'] = df['time_bin'].clip(0, n_bins-1)

# 2. Create a grid: rows = functions, cols = time bins
functions = df['function'].unique()
grid = np.zeros((len(functions), n_bins), dtype=int)

func_idx = {f: i for i, f in enumerate(functions)}
for _, row in df.iterrows():
    i = func_idx[row['function']]
    j = row['time_bin']
    grid[i, j] = 1  # function called

# 3. Plot as binary image
plt.figure(figsize=(12, 8))
plt.imshow(grid, aspect='auto', cmap=plt.cm.coolwarm_r)  # yellow=1, blue=0
plt.axis('off')
plt.show()
```
