# Ninety years later, the redlining map still predicts who carries the burden

Companion piece to `dc_ej_predictors.ipynb`: that piece found that merascope's 21 physical/infrastructure siting factors barely explain present-day community burden (R\u00b2=0.25, real but modest). This piece checks a specific historical candidate for the other three-quarters -- 1930s HOLC "residential security" redlining maps, digitized by the University of Richmond's [Mapping Inequality](https://dsl.richmond.edu/panorama/redlining/) project (data via the [source.coop mirror](https://source.coop/cboettig/mappinginequality), CC BY-NC-SA 4.0 -- non-commercial use, attribution required, matches this campaign's non-commercial LinkedIn context).

**Not a novel finding.** Redlining's link to present-day disparity is well-established in the literature (e.g. Mitchell & Franco 2018, Nardone et al. 2020 on urban heat). The honest framing here is *does Washington's own data show the same pattern documented nationally elsewhere* -- a replication check on real WA data, not a discovery.

**Real coverage constraint, not a choice:** HOLC only surveyed larger cities in the 1930s. Washington has exactly 3: Seattle, Tacoma, Spokane. Rural WA and the merascope data-center clusters (Quincy, the desert campuses) were never surveyed and have no redlining data -- this piece can't be statewide the way the last one was.

```python
from pathlib import Path

import geopandas as gpd
import numpy as np
import pandas as pd

HOLC_PATH = Path("data/mappinginequality_wa.geojson")
holc = gpd.read_file(HOLC_PATH)
print(f"{len(holc)} WA HOLC polygons: {holc['city'].value_counts().to_dict()}")
print(f"grades: {holc['grade'].value_counts().to_dict()}")

GRADE_COLOR = {"A": "#76a865", "B": "#7cb5bd", "C": "#ffff00", "D": "#d9838d"}
CITIES = ["Seattle", "Tacoma", "Spokane"]
```

## Matching resolution, not reusing the ZCTA score

HOLC polygons are neighborhood-scale -- much finer than merascope's ZCTA grid. Several differently-graded HOLC areas sit inside a single Seattle ZCTA, so reusing the ZCTA-level `ej_score` from the companion piece would blur grades together (the same resolution-mismatch mistake flagged and avoided there). Instead this reuses the tract-level `demog_index` this campaign already built (`dc_ej_predictors.ipynb`'s MOE-check cache) and overlays it directly against the real HOLC polygons -- same population-weighted polygon-overlay technique, just against a finer target than a ZCTA.

```python
acs = pd.read_csv("data/acs_demog_moe_tract_wa.csv", dtype={"GEOID": str})
tracts = gpd.read_file("../../merascope/data/WA/raw/tracts.geojson")[["GEOID", "geometry"]].merge(acs, on="GEOID")

holc_m = holc.to_crs(3857)
tracts_m = tracts.set_geometry("geometry").to_crs(3857)
tracts_m["tract_area"] = tracts_m.geometry.area

overlay = gpd.overlay(holc_m[["area_id", "city", "grade", "geometry"]], tracts_m, how="intersection", keep_geom_type=False)
overlay["piece_area"] = overlay.geometry.area
overlay = overlay.merge(tracts_m[["GEOID", "tract_area"]], on="GEOID", suffixes=("", "_full"))
overlay["pop_weight"] = overlay["tract_pop"] * (overlay["piece_area"] / overlay["tract_area"])


def agg(g):
    w = g["pop_weight"]
    if w.sum() <= 0:
        return pd.Series({"demog_index": np.nan, "demog_index_cv": np.nan, "n_tracts": g["GEOID"].nunique()})
    di = np.average(g["demog_index"], weights=w)
    cv = np.average(g["demog_index_cv"].fillna(g["demog_index_cv"].median()), weights=w)
    return pd.Series({"demog_index": di, "demog_index_cv": cv, "n_tracts": g["GEOID"].nunique()})


holc_agg = overlay.groupby("area_id").apply(agg, include_groups=False).reset_index()
holc = holc.merge(holc_agg, on="area_id", how="left")

n_missing = holc["demog_index"].isna().sum()
print(f"{n_missing} of {len(holc)} HOLC polygons got no tract match (unpopulated -- water/industrial)")

holc_valid = holc.dropna(subset=["demog_index", "grade"]).copy()
unique_tracts = overlay["GEOID"].nunique()
print(f"n={len(holc_valid)} graded polygons; median {holc_valid['n_tracts'].median():.0f} tracts/polygon; "
      f"{unique_tracts} unique tracts touched overall")
print(f"-- not {len(holc_valid)} fully independent observations: nearby polygons often share a tract. "
      f"Reported honestly, not treated as {len(holc_valid)} independent draws.")

print(f"\nreliability: median CV = {holc_valid['demog_index_cv'].median()*100:.1f}%, "
      f"{(holc_valid['demog_index_cv']>0.30).mean()*100:.1f}% of polygons exceed CV>30% "
      f"(better than the ZCTA check's 14% -- smaller polygons average over fewer, but the tract data itself is the same source)")
```

## Does the 1930s grade actually predict today's burden?

```python
from scipy import stats

GRADE_RANK = {"A": 1, "B": 2, "C": 3, "D": 4}
holc_valid["grade_rank"] = holc_valid["grade"].map(GRADE_RANK).astype(float)

groups = [holc_valid[holc_valid["grade"] == g]["demog_index"].to_numpy(dtype=float) for g in "ABCD"]
kw = stats.kruskal(*groups)
rho, p = stats.spearmanr(holc_valid["grade_rank"].to_numpy(dtype=float), holc_valid["demog_index"].to_numpy(dtype=float))
print(f"Kruskal-Wallis across A/B/C/D: H={kw.statistic:.1f}, p={kw.pvalue:.2e}")
print(f"Spearman rho (grade rank vs demog_index) = {rho:.3f}, p = {p:.2e}")
assert p < 0.01, f"redlining grade doesn't clear significance here (p={p:.3f}) -- don't claim a real link"

print("\nper city (does it hold everywhere, or just one place?):")
for city in CITIES:
    sub = holc_valid[holc_valid["city"] == city]
    rho_c, p_c = stats.spearmanr(sub["grade_rank"].to_numpy(dtype=float), sub["demog_index"].to_numpy(dtype=float))
    means = sub.groupby("grade")["demog_index"].mean()
    print(f"  {city}: n={len(sub)}, rho={rho_c:.3f} p={p_c:.3f}, means A->D: "
          f"{[round(float(means.get(g, float('nan'))), 3) for g in 'ABCD']}")

d_mean = holc_valid[holc_valid["grade"] == "D"]["demog_index"].mean()
a_mean = holc_valid[holc_valid["grade"] == "A"]["demog_index"].mean()
assert d_mean > a_mean, "formerly redlined areas don't actually read higher-burden than 'Best' areas -- claim doesn't hold"
print(f"\nD (Hazardous/redlined) vs A (Best): {d_mean:.3f} vs {a_mean:.3f} -- {d_mean/a_mean:.2f}x")
print("Note: Tacoma's C actually reads slightly higher than its D (0.242 vs 0.208) -- the overall "
      "pattern holds (both C and D sit well above A/B everywhere), but it's not a perfectly monotonic "
      "staircase in every city. Reported as-is, not smoothed over.")
```

```python
import matplotlib.pyplot as plt

BAR_COLOR = {"A": "#4d7a3a", "B": "#3a7a86", "C": "#c9a800", "D": "#b8425a"}
fig, ax = plt.subplots(figsize=(9, 6))
width = 0.2
x = np.arange(len(CITIES))

for i, g in enumerate("ABCD"):
    means, errs = [], []
    for city in CITIES:
        sub = holc_valid[(holc_valid["city"] == city) & (holc_valid["grade"] == g)]["demog_index"]
        means.append(sub.mean() if len(sub) else np.nan)
        errs.append(sub.std() / np.sqrt(len(sub)) if len(sub) > 1 else 0)
    ax.bar(x + (i - 1.5) * width, means, width, yerr=errs, label=g, color=BAR_COLOR[g], capsize=3)

ax.set_xticks(x)
ax.set_xticklabels(CITIES)
ax.set_ylabel("demog_index (poverty + minority rate, present-day ACS)\nhigher = more burden")
ax.set_title(
    f"Present-day burden by 1930s HOLC grade, all 3 WA cities surveyed\n"
    f"Spearman rho={rho:.2f}, p={p:.1e}, n={len(holc_valid)} -- error bars = SEM",
    fontsize=12,
)
ax.legend(title="HOLC grade", loc="upper left")
for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)
fig.tight_layout()
fig.savefig("dc_ej_redlining_bar.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
import json

from matplotlib.patches import Polygon
from matplotlib.collections import PatchCollection


def rings(geometry):
    coords = geometry["coordinates"]
    polys = [coords] if geometry["type"] == "Polygon" else coords
    for poly in polys:
        yield np.array(poly[0])


vmin, vmax = holc_valid["demog_index"].min(), holc_valid["demog_index"].max()
fig, axes = plt.subplots(2, 3, figsize=(15, 11))

for col, city in enumerate(CITIES):
    sub = holc_valid[holc_valid["city"] == city]
    feats = json.loads(sub.to_json())["features"]
    all_lats = np.concatenate([r[:, 1] for f in feats for r in rings(f["geometry"])])
    lat_aspect = 1 / np.cos(np.radians(all_lats.mean()))

    ax_top = axes[0, col]
    patches_top, colors_top = [], []
    for f in feats:
        for ring in rings(f["geometry"]):
            patches_top.append(Polygon(ring, closed=True))
            colors_top.append(GRADE_COLOR[f["properties"]["grade"]])
    coll_top = PatchCollection(patches_top, facecolor=colors_top, edgecolor="white", linewidth=0.3)
    ax_top.add_collection(coll_top)
    ax_top.autoscale_view()
    ax_top.set_aspect(lat_aspect)
    ax_top.axis("off")
    ax_top.set_title(f"{city}, 1930s HOLC grade", fontsize=11)

    ax_bot = axes[1, col]
    patches_bot, values_bot = [], []
    for f in feats:
        for ring in rings(f["geometry"]):
            patches_bot.append(Polygon(ring, closed=True))
            values_bot.append(f["properties"]["demog_index"])
    coll_bot = PatchCollection(patches_bot, cmap="YlOrRd")
    coll_bot.set_array(np.array(values_bot))
    coll_bot.set_clim(vmin, vmax)
    coll_bot.set_edgecolor("white")
    coll_bot.set_linewidth(0.3)
    ax_bot.add_collection(coll_bot)
    ax_bot.autoscale_view()
    ax_bot.set_aspect(lat_aspect)
    ax_bot.axis("off")
    ax_bot.set_title(f"{city}, present-day burden\n(same footprint)", fontsize=11)

handles = [plt.Rectangle((0, 0), 1, 1, facecolor=GRADE_COLOR[g], edgecolor="white") for g in "ABCD"]
labels = ["A - Best", "B - Still Desirable", "C - Definitely Declining", "D - Hazardous"]
fig.legend(handles, labels, loc="upper center", ncol=4, bbox_to_anchor=(0.5, 0.985), fontsize=9, frameon=False)

sm = plt.cm.ScalarMappable(cmap="YlOrRd", norm=plt.Normalize(vmin=vmin, vmax=vmax))
sm.set_array([])
cbar = fig.colorbar(sm, ax=axes[1, :], orientation="horizontal", fraction=0.05, pad=0.05, aspect=40)
cbar.set_label("demog_index (poverty + minority rate, present-day ACS)")

fig.suptitle("Ninety years later, the 1930s redlining map still predicts who carries the burden today", fontsize=13, y=1.02)
fig.savefig("dc_ej_redlining_map.png", dpi=150, bbox_inches="tight")
plt.show()
```

## Source & caption notes

- Data: [Mapping Inequality](https://dsl.richmond.edu/panorama/redlining/) (University of Richmond Digital Scholarship Lab), 1930s HOLC "residential security" grades, via the [source.coop GeoParquet mirror](https://source.coop/cboettig/mappinginequality). CC BY-NC-SA 4.0 -- credit the source, non-commercial use only (fine for this campaign, not for merascope-the-product). 166 WA polygons across the only 3 WA cities HOLC surveyed in the 1930s: Seattle (62), Tacoma (49), Spokane (55).
- Present-day burden: the same tract-level ACS `demog_index` (poverty rate + minority rate, averaged) that `ej_score` is built from in merascope's pipeline -- but matched at HOLC-polygon resolution via a real population-weighted polygon overlay, not the coarser ZCTA score.
- **Not a novel finding.** Redlining's link to present-day disparity is well-documented nationally. This checks whether WA's own cities show the same pattern -- they do.
- **Result: yes, and it's not close.** Spearman rho=0.47 (n=159, p=4.6e-10) between 1930s grade and today's demog_index; Kruskal-Wallis across the 4 grades p=1.6e-8. Formerly redlined ("D -- Hazardous") areas average 1.84x the present-day poverty+minority rate of "A -- Best" areas. Holds in all 3 cities individually (Seattle rho=0.39 p=0.002, Tacoma rho=0.42 p=0.004, Spokane rho=0.59 p<0.001) -- not one city driving the whole result.
- **Honest wrinkle, not smoothed over:** Tacoma's C grade actually reads slightly higher-burden than its D grade (0.242 vs 0.208) -- the overall staircase (A<B<C<D) isn't perfectly monotonic in every city, even though the broad C/D-above-A/B pattern holds everywhere.
- **Independence caveat:** 159 graded polygons touch only 263 unique census tracts (median 3 tracts per polygon) -- nearby same-grade polygons often share a tract, so this isn't 159 fully independent observations. The effect size and per-city replication make this unlikely to be an artifact, but it's not claimed as more statistically independent than it is.
- Reliability: median per-polygon CV 15.5%, 5.5% of polygons exceed the Census's own CV>30% caution threshold -- better than the companion ZCTA piece (14% failing), because HOLC polygons average over multiple tracts (median 3) rather than one ZCTA's worth.
