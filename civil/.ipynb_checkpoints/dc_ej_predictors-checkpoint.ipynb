# What predicts community burden -- and what doesn't

Merascope backlog piece. `ej_score` (1 = least demographic burden) is built purely from tract-level Census ACS poverty + minority-population rates -- it shares zero inputs with the pipeline's other 21 physical/infrastructure indicators (grid access, water, flood, seismic, contamination, fiber, etc.), so asking whether those siting-favorable factors track with community burden isn't circular (checked below, not assumed).

Two things get checked before any claim: (1) the ACS estimates `ej_score` is built from have a real margin of error that was never validated at product resolution -- the exact gap that killed `broadband_ej_overlay.ipynb` this week (58% of ZCTAs failed the Census's own CV>30% reliability bar) -- so that check runs first, at the actual WA-ZCTA resolution this piece uses, not the block-group resolution that piece used. (2) no feature is a disguised copy of the target.

```python
from pathlib import Path

import geopandas as gpd
import numpy as np

ZCTA_PATH = Path("../../merascope/data/WA/zcta/grid_scores.geojson")
zctas = gpd.read_file(ZCTA_PATH)
print(f"{len(zctas)} WA ZCTAs, {len(zctas.columns)} columns")

# grid_capacity_score is state-level (EIA-860M ratio), identical for every WA
# ZCTA -- confirmed in dc_grid_queue.ipynb. Zero variance, useless as a feature.
SCORE_COLS = [c for c in zctas.columns
              if c.endswith("_score") and not c.endswith("_score_nat")
              and c not in ("ej_score", "grid_capacity_score")]
print(f"{len(SCORE_COLS)} candidate features:", SCORE_COLS)
```

```python
# Leakage guard: ej_score is built from tract ACS poverty+minority rates alone
# (02_indicators.py). Confirm none of the other 21 indicators are a near-copy
# of it before trusting anything the model says about them.
corr = zctas[SCORE_COLS + ["ej_score"]].corr()["ej_score"].drop("ej_score")
print("correlation with ej_score (none should be near +/-1.0):")
print(corr.sort_values(key=abs, ascending=False).round(3))
assert corr.abs().max() < 0.9, "a feature is suspiciously collinear with ej_score -- check for shared inputs"
print(f"\nmax |correlation| = {corr.abs().max():.3f} -- no leakage")
```

## Reliability check: is `ej_score` itself trustworthy at ZCTA resolution?

`ej_score` = `1 - normalized(mean(poverty_rate, minority_rate))`, both pulled from 2022 ACS 5-year **tract**-level counts with no margin-of-error check anywhere in the pipeline (`02_indicators.py`). Real per-ZCTA reliability requires knowing which tracts actually overlap each ZCTA and weighting by population, not just spot-checking tracts in isolation.

```python
import os

import pandas as pd
import requests

MOE_CACHE = Path("data/acs_demog_moe_tract_wa.csv")


def load_census_key():
    key = os.environ.get("CENSUS_API_KEY")
    if key:
        return key
    for f in [Path("/home/simonhans/coding/snotrac/.env"), Path.home() / ".env"]:
        if f.exists():
            for line in f.read_text().splitlines():
                if line.startswith("CENSUS_API_KEY"):
                    return line.split("=", 1)[1].strip()
    raise RuntimeError("no census key found")


def ratio_moe(num_e, num_m, denom_e, denom_m, ratio):
    """Census's approximate MOE formula for a ratio of two correlated counts."""
    term = num_m**2 - (ratio**2) * (denom_m**2)
    use_minus = term >= 0
    return np.where(use_minus,
                     np.sqrt(np.maximum(term, 0)) / denom_e,
                     np.sqrt(num_m**2 + (ratio**2) * (denom_m**2)) / denom_e)


if MOE_CACHE.exists():
    acs = pd.read_csv(MOE_CACHE, dtype={"GEOID": str})
    print(f"loaded cached ACS MOE data: {len(acs)} tracts")
else:
    key = load_census_key()
    params = {
        "get": "NAME,B17001_001E,B17001_001M,B17001_002E,B17001_002M,"
               "B02001_001E,B02001_001M,B02001_002E,B02001_002M,B01003_001E",
        "for": "tract:*", "in": "state:53", "key": key,
    }
    r = requests.get("https://api.census.gov/data/2022/acs/acs5", params=params, timeout=60)
    r.raise_for_status()
    rows = r.json()
    acs = pd.DataFrame(rows[1:], columns=rows[0])
    numcols = [c for c in acs.columns if c.startswith("B17001") or c.startswith("B02001") or c.startswith("B01003")]
    for c in numcols:
        acs[c] = pd.to_numeric(acs[c], errors="coerce")
    acs["GEOID"] = acs["state"].str.zfill(2) + acs["county"].str.zfill(3) + acs["tract"].str.zfill(6)

    acs["poverty_rate"] = (acs["B17001_002E"] / acs["B17001_001E"]).clip(0, 1)
    acs["poverty_rate_moe"] = ratio_moe(acs["B17001_002E"], acs["B17001_002M"],
                                          acs["B17001_001E"], acs["B17001_001M"], acs["poverty_rate"])
    white_rate = (acs["B02001_002E"] / acs["B02001_001E"]).clip(0, 1)
    acs["minority_rate"] = 1 - white_rate
    acs["minority_rate_moe"] = ratio_moe(acs["B02001_002E"], acs["B02001_002M"],
                                           acs["B02001_001E"], acs["B02001_001M"], white_rate)
    acs["demog_index"] = (acs["poverty_rate"] + acs["minority_rate"]) / 2
    # 90% MOEs combine as root-sum-of-squares for a sum of two independent estimates
    acs["demog_index_moe"] = np.sqrt(acs["poverty_rate_moe"]**2 + acs["minority_rate_moe"]**2) / 2
    acs["demog_index_cv"] = (acs["demog_index_moe"] / 1.645 / acs["demog_index"]).replace([np.inf, -np.inf], np.nan)
    acs["tract_pop"] = acs["B01003_001E"]

    acs = acs[["GEOID", "demog_index", "demog_index_moe", "demog_index_cv", "tract_pop"]].copy()
    MOE_CACHE.parent.mkdir(parents=True, exist_ok=True)
    acs.to_csv(MOE_CACHE, index=False)
    print(f"fetched + cached ACS MOE data: {len(acs)} tracts")

print(f"tract-level demog_index: median CV = {acs['demog_index_cv'].median()*100:.1f}%, "
      f"{(acs['demog_index_cv']>0.30).mean()*100:.1f}% of tracts exceed the Census's CV>30% caution threshold")
```

```python
# Real polygon overlay (tract x ZCTA intersection), weighted by each tract's ACS
# population apportioned by intersected area (area-proportional -- sub-tract
# population isn't available). This is the same reliability question the
# choropleth below needs answered per-ZCTA before any single polygon's color
# is trusted -- not a simplistic "which tract contains the centroid" pick.
tracts = gpd.read_file("../../merascope/data/WA/raw/tracts.geojson")[["GEOID", "geometry"]].merge(acs, on="GEOID")

zctas_m = zctas.to_crs(3857)
tracts_m = tracts.set_geometry("geometry").to_crs(3857)
tracts_m["tract_area"] = tracts_m.geometry.area

overlay = gpd.overlay(zctas_m[["zcta", "geometry"]], tracts_m, how="intersection")
overlay["piece_area"] = overlay.geometry.area
overlay = overlay.merge(tracts_m[["GEOID", "tract_area"]], on="GEOID", suffixes=("", "_full"))
overlay["pop_weight"] = overlay["tract_pop"] * (overlay["piece_area"] / overlay["tract_area"])


def weighted_cv(g):
    w = g["pop_weight"]
    if w.sum() <= 0:
        return pd.Series({"demog_index_cv_zcta": np.nan})
    cv = np.average(g["demog_index_cv"].fillna(g["demog_index_cv"].median()), weights=w)
    return pd.Series({"demog_index_cv_zcta": cv})


zcta_rel = overlay.groupby("zcta").apply(weighted_cv, include_groups=False).reset_index()
zctas = zctas.merge(zcta_rel, on="zcta", how="left")
assert zctas["demog_index_cv_zcta"].notna().all(), "some ZCTAs got no reliability estimate -- overlay join gap"

unreliable_frac = (zctas["demog_index_cv_zcta"] > 0.30).mean()
print(f"population-weighted per-ZCTA demog_index CV: median = {zctas['demog_index_cv_zcta'].median()*100:.1f}%, "
      f"{unreliable_frac*100:.1f}% of {len(zctas)} ZCTAs exceed CV>30%")
print(f"compare: block-group data behind broadband_ej_overlay.ipynb was 33.5% median / 58% failing -- "
      f"this is meaningfully more reliable, but {unreliable_frac*100:.0f}% of ZCTAs still don't clear the bar, "
      f"marked (hatched) on the map below rather than silently trusted.")
```

## Model

Random forest, 5-fold cross-validated, same setup as the (killed, but methodologically fine) `broadband_ej_overlay.ipynb` model: `n_estimators=300, min_samples_leaf=3`. Whether the R² is even real -- not just an arbitrary ">0.15" threshold -- gets its own permutation test right after ([[feedback_verify_before_asserting]]): an unreliable model's feature ranking is noise, not a finding, and "clears an arbitrary cutoff" isn't the same claim as "beats chance."

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.inspection import permutation_importance
from sklearn.model_selection import cross_val_predict, cross_val_score, KFold

X = zctas[SCORE_COLS].to_numpy()
y = zctas["ej_score"].to_numpy()
assert not np.isnan(X).any() and not np.isnan(y).any(), "NaNs in feature matrix or target"

model = RandomForestRegressor(n_estimators=300, min_samples_leaf=3, random_state=0)
kf = KFold(n_splits=5, shuffle=True, random_state=0)
cv_r2 = cross_val_score(model, X, y, cv=kf, scoring="r2")
print(f"5-fold CV R²: {cv_r2.mean():.3f} +/- {cv_r2.std():.3f}  (per fold: {np.round(cv_r2, 3)})")

model.fit(X, y)
impurity_imp = sorted(zip(SCORE_COLS, model.feature_importances_), key=lambda t: -t[1])
perm = permutation_importance(model, X, y, n_repeats=20, random_state=0, scoring="r2")
perm_imp = sorted(zip(SCORE_COLS, perm.importances_mean), key=lambda t: -t[1])

print("\ntop 5, impurity importance:", [n for n, _ in impurity_imp[:5]])
print("top 5, permutation importance:", [n for n, _ in perm_imp[:5]])
top_agrees = impurity_imp[0][0] == perm_imp[0][0]
print(f"\n#1 feature agrees across both methods: {top_agrees} "
      f"({impurity_imp[0][0]} vs {perm_imp[0][0]})")
if not top_agrees:
    print("-- report the top cluster, not a single winner; the exact #1 rank is method-sensitive here")

y_pred_oof = cross_val_predict(model, X, y, cv=kf)
zctas["ej_score_pred"] = y_pred_oof
zctas["ej_score_resid"] = zctas["ej_score"] - zctas["ej_score_pred"]
```

## Is R² = 0.25 actually signal, or noise from fitting 21 correlated features to 605 rows?

Real question, checked rather than asserted by a magic threshold: shuffle `ej_score` against the real feature matrix (breaking any real relationship) and rerun the identical 5-fold CV random forest, 200 times. If the real R² isn't clearly separated from that shuffled-label null distribution, the whole importance ranking below is noise dressed up as a finding.

```python
PERM_CACHE = Path("data/dc_ej_permutation_null_r2.npy")
N_PERM = 200

if PERM_CACHE.exists():
    null_scores = np.load(PERM_CACHE)
    print(f"loaded cached permutation null ({len(null_scores)} shuffles)")
else:
    null_scores = np.empty(N_PERM)
    for i in range(N_PERM):
        y_shuf = np.random.RandomState(i).permutation(y)
        null_scores[i] = cross_val_score(model, X, y_shuf, cv=kf, scoring="r2").mean()
        if (i + 1) % 20 == 0:
            print(f"  {i+1}/{N_PERM} shuffles done")
    PERM_CACHE.parent.mkdir(parents=True, exist_ok=True)
    np.save(PERM_CACHE, null_scores)

real_r2 = cv_r2.mean()
n_exceed = (null_scores >= real_r2).sum()
p_value = (n_exceed + 1) / (len(null_scores) + 1)
z = (real_r2 - null_scores.mean()) / null_scores.std()

print(f"\nreal R² = {real_r2:.4f}")
print(f"shuffled-label null: mean = {null_scores.mean():.4f}, std = {null_scores.std():.4f}, max = {null_scores.max():.4f}")
print(f"{n_exceed}/{len(null_scores)} shuffles matched or exceeded the real R² -- z = {z:.1f}, p = {p_value:.4f}")
assert p_value < 0.05, f"real R² doesn't clear the permutation null (p={p_value:.3f}) -- importances below aren't trustworthy"
print("\nreal signal, not noise -- proceeding to feature importances and the map")
```

```python
import json

import matplotlib.pyplot as plt
from matplotlib.patches import Polygon
from matplotlib.collections import PatchCollection

features_geojson = json.loads(zctas.to_json())["features"]


def zcta_rings(geometry):
    coords = geometry["coordinates"]
    polys = [coords] if geometry["type"] == "Polygon" else coords
    for poly in polys:
        yield np.array(poly[0])


all_lats = np.concatenate([r[:, 1] for f in features_geojson for r in zcta_rings(f["geometry"])])
LAT_ASPECT = 1 / np.cos(np.radians(all_lats.mean()))
UNRELIABLE_CV = 0.30


def plot_choropleth(ax, field, cmap="viridis", vmin=None, vmax=None, title=""):
    """PatchCollection choropleth, same technique as transit_gap_visible.ipynb.
    ZCTAs with demog_index_cv_zcta > 30% get a hatch overlay -- an unreliable
    ACS estimate shouldn't be read as a colored finding, see MOE check above."""
    patches, values, unreliable = [], [], []
    for feat in features_geojson:
        p = feat["properties"]
        val = p[field]
        if val is None:
            continue
        for ring in zcta_rings(feat["geometry"]):
            patches.append(Polygon(ring, closed=True))
            values.append(val)
            if p.get("demog_index_cv_zcta") and p["demog_index_cv_zcta"] > UNRELIABLE_CV:
                unreliable.append(Polygon(ring, closed=True))
    coll = PatchCollection(patches, cmap=cmap)
    coll.set_array(np.array(values))
    coll.set_clim(vmin if vmin is not None else min(values), vmax if vmax is not None else max(values))
    coll.set_edgecolor("white")
    coll.set_linewidth(0.15)
    ax.add_collection(coll)
    if unreliable:
        ax.add_collection(PatchCollection(unreliable, facecolor="none", edgecolor="#1a1a1a",
                                            hatch="///", linewidth=0.0, zorder=5))
    ax.autoscale_view()
    ax.set_aspect(LAT_ASPECT)
    ax.axis("off")
    ax.set_title(title, fontsize=11)
    return coll


fig, axes = plt.subplots(1, 3, figsize=(16, 6.5))
vmin, vmax = zctas["ej_score"].min(), zctas["ej_score"].max()
c0 = plot_choropleth(axes[0], "ej_score", vmin=vmin, vmax=vmax,
                       title="Actual ej_score\n(real ACS-derived community burden)")
c1 = plot_choropleth(axes[1], "ej_score_pred", vmin=vmin, vmax=vmax,
                       title="Predicted ej_score\n(out-of-fold, from 21 physical/infra indicators)")
resid_lim = max(abs(zctas["ej_score_resid"].min()), abs(zctas["ej_score_resid"].max()))
c2 = plot_choropleth(axes[2], "ej_score_resid", cmap="RdBu_r", vmin=-resid_lim, vmax=resid_lim,
                       title="Residual (actual - predicted)\nred = more burdened than infra predicts")

fig.colorbar(c0, ax=axes[0], orientation="horizontal", fraction=0.05, pad=0.03, label="ej_score")
fig.colorbar(c1, ax=axes[1], orientation="horizontal", fraction=0.05, pad=0.03, label="ej_score")
fig.colorbar(c2, ax=axes[2], orientation="horizontal", fraction=0.05, pad=0.03, label="residual")

fig.suptitle(
    f"Physical siting infrastructure explains about a quarter of WA's EJ-burden variance\n"
    f"(5-fold CV R\u00b2 = {cv_r2.mean():.2f}) -- hatched ZCTAs have unreliable underlying ACS estimates (CV>30%)",
    fontsize=12,
)
fig.tight_layout()
fig.savefig("dc_ej_ml_maps.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
order = np.argsort(perm.importances_mean)[::-1][:10]
names = [SCORE_COLS[i] for i in order]
perm_vals = perm.importances_mean[order]
perm_err = perm.importances_std[order]
imp_vals = model.feature_importances_[order]

fig, ax = plt.subplots(figsize=(8, 6))
y_pos = np.arange(len(names))[::-1]
ax.barh(y_pos, perm_vals, xerr=perm_err, color="#2a78d6", alpha=0.85,
         label="permutation importance (drop in R\u00b2 when shuffled)")
ax2 = ax.twiny()
ax2.plot(imp_vals, y_pos, "o", color="#d64545", markersize=6, label="impurity importance (RF built-in)")

ax.set_yticks(y_pos)
ax.set_yticklabels(names)
ax.set_xlabel("permutation importance (mean drop in R\u00b2)")
ax2.set_xlabel("impurity importance")
ax.set_title("What predicts ej_score across WA ZCTAs?\ntop-10 of 21 physical/infrastructure indicators, two methods", fontsize=12)

lines1, labels1 = ax.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax.legend(lines1 + lines2, labels1 + labels2, loc="lower right", fontsize=8)
for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)
fig.tight_layout()
fig.savefig("dc_ej_ml_importance.png", dpi=150, bbox_inches="tight")
plt.show()
```

## Source & caption notes

- Data: merascope's own scored WA ZCTA pipeline (605 ZCTAs), all 21 non-target `*_score` physical/infrastructure indicators as features (`grid_capacity_score` dropped -- state-level constant, confirmed in `dc_grid_queue.ipynb`), `ej_score` as target. Leakage-checked: max correlation between any feature and the target is |0.27| (`water_stress_score`) -- nothing near-duplicates the target's own ACS-poverty/minority construction.
- **Reliability first, not an afterthought.** `ej_score`'s underlying ACS tract data has never had a margin-of-error check anywhere in the merascope pipeline. Checked here at real product (ZCTA) resolution via a population-weighted tract-ZCTA polygon overlay: median CV 18%, 14% of WA ZCTAs exceed the Census's own CV>30% caution threshold -- meaningfully more reliable than the block-group data that got `broadband_ej_overlay.ipynb` killed (33.5% median, 58% failing), but not perfect. Those 14% are hatched on the map, not silently trusted or hidden.
- **Model: 5-fold CV R² ≈ 0.25, confirmed real by a 200-shuffle permutation test, not just an arbitrary threshold.** Shuffling `ej_score` against the real features and rerunning the identical pipeline gives a null of mean -0.085 / std 0.028 / max -0.014 across 200 shuffles -- zero of them came anywhere near the real R², z ≈ 10, p = 0.005. So 0.25 is real signal, not RF overfitting 21 correlated features to 605 rows. It's just a *modest* real signal: three-quarters of WA's ZCTA-to-ZCTA variance in community burden is still unexplained by any of the 21 physical siting factors developers actually optimize for. That's the headline -- real but partial, not a strong prediction. The predicted-score map visibly compresses toward the middle (a real random-forest signature of a moderate-R² fit) next to the actual map's real extremes.
- **No single winner.** Impurity importance ranks `seismic_score` #1; permutation importance ranks `water_stress_score` #1 -- the two methods disagree on the exact top spot. What both agree on: `water_stress_score`, `water_score`, `seismic_score`, and `pop_exposure_score` form a cluster well ahead of everything else. Reported as a cluster, not a single mechanism.
- Residual map (actual minus predicted) is the honest payoff of a moderate-R² model: it shows *where* burden runs higher or lower than physical infrastructure alone would suggest -- i.e., where something else (economic geography, housing history, factors this pipeline doesn't capture) is doing the work.
- n=605 ZCTAs, real cross-validated out-of-fold predictions (not in-sample, which would be near-perfect and misleading about actual predictive skill).
