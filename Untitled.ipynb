```python
import functools
import time

call_log = []

def listen(func):
    """Decorator to record calls to a function."""
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        t = time.time()
        call_log.append({
            "function": func.__name__,
            "time": t,
            "args": args,
            "kwargs": kwargs
        })
        return func(*args, **kwargs)
    return wrapper
```

```python
@listen
def foo(x):
    return x**2

@listen
def bar(y):
    return foo(y) + 1

bar(3)
bar(7)

print(call_log)
```

```python
import sys
import time

trace_log = []

def tracer(frame, event, arg):
    if event == "call":
        func_name = frame.f_code.co_name
        filename = frame.f_code.co_filename
        timestamp = time.time()
        trace_log.append({
            "function": func_name,
            "file": filename,
            "time": timestamp
        })
    return tracer

sys.settrace(tracer)

# ---- run your program here ----
def demo():
    return sum([i*i for i in range(5)])

demo()

sys.settrace(None)
print(trace_log[:5])
```

```python
from collections import defaultdict

call_counts = defaultdict(int)

def listen_count(func):
    def wrapper(*args, **kwargs):
        call_counts[func.__name__] += 1
        return func(*args, **kwargs)
    return wrapper
```
