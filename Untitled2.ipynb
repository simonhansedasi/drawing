```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.collections import LineCollection

fig, ax = plt.subplots(figsize = (10,10))
ax.set_aspect('equal')
ax.set_facecolor("#0b3d91")
plt.axis('off')

# --- Parameters ---
center = np.array([0.5, 0.5])
pos = center.copy()

steps = 2000
length = 0.0025           # base step size
radius_growth = 0.0008    # radial drift per step

# Color progression
colors = np.linspace(0, 1, steps)

# Initial angle
angle = 0.0

# Angle dynamics
angle_change_base = 0.15
angle_change_wave = 0.08
angle_wave_freq = 0.02

# Store points
points = [pos.copy()]

# Outward spiral loop
for i in range(steps):
    # Gradually increase radius
    r = i * radius_growth
    
    # Update angle using sin modulation
    angle += angle_change_base + np.sin(i * angle_wave_freq) * angle_change_wave
    
    # Next point in polar coordinates
    x_new = center[0] + r * np.cos(angle)
    y_new = center[1] + r * np.sin(angle)
    
    points.append([x_new, y_new])

points = np.array(points)

# Create gradient-colored segments
segments = np.stack([points[:-1], points[1:]], axis=1)
lc = LineCollection(segments, cmap='viridis', norm=plt.Normalize(0, 1))
lc.set_array(colors)
lc.set_linewidth(2)

ax.add_collection(lc)
ax.set_xlim(-0.1, 1.1)
ax.set_ylim(-0.1, 1.1)

plt.show()
```
